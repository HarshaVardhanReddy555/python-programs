```python
#2.given year is leapyear or not
year=int(input("enter year "))
if(year%4==0 and year%100!=0) or ( year%400==0):
  print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")
```

```python
#3.example code on match-case
day=int(input("enter a number"))
match day:
  case 1:
    print("monday")
  case 2:
    print("tuesday")
  case 3:
    print("wednesday")
  case 4:
    print("thuesday")
  case 5:
    print("friday")
  case 6:
    print("saturday")
  case 7:
    print("sunday")
  case _:
    print("unvalid input")
```

LOOPS IN PYTHON

WHILE LOOP

```python

#1.printing 1 to 10
i=1
while(i<=10):
  print( i ,end=" ")
  i+=1
```

```python
#2.printing n to 1 numbers
n=int(input("enter value where to start print "))
while(n>=1):
  print(n,end=" ")
  n-=1
```

```python
#3.printing 1 to n even numbers only
n=int(input("enter range"))
i=1
while(i<=n):
  if(i%2==0):
    print(i,end=" ")
  i+=1
```

FOR LOOP

```python
#1.printing 1 to 10
for i in range(1,10):
  print(i,end=" ")
```

```python
#2.printing n to 1 numbers
n=int(input("enter value where to start print "))
for i in range(n,0,-1):
  print(i,end=" ")
```

```python
#3.printing 1 to n even numbers only
n=int(input("enter range"))
for i in range(0,n,2):
  print(i,end=" ")
```

```python
#4.program that prints table upto 10 multiple
n=int(input("enter a number"))
for i in range(1,11):
  print(f"{n} * {i} = {n*i}")
```

```python
#program that finds sum of thr digits of the given number
n=int(input("enter a number "))
sum=0
while(n>0):
  r=n%10
  sum=sum+r
  n=n//10
print(f"sum of digits in number is {sum}")
```

```python
#reverse of a number
n=int(input("enter a number "))
rev=0
while(n>0):
  r=n%10
  rev=rev*10+r
  n=n//10
print(f"reverse of number is {rev}")
```

```python
#to check if the number is prime number or not
n=int(input("enter a number "))
for i in range(2,n):
  if(n%i==0):
    print(f"{n} is not a prime number")
    break
  else:
    print(f"{n} is a prime number")


```

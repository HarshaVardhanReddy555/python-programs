```python
#functions that returns multiple values
def multiple_function(*num):
  a=sum(num)
  b=len(num)
  avg=a/b
  return a,b,avg
res_sum,res_len,res_avg=multiple_function(1,4,5,3,8,3,8,7)
print(res_sum,res_len,res_avg)
```

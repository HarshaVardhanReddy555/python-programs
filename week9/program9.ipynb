```python
#program on '"w" mode
file=open("hvr.txt","w")
file.write("harsha vardhan reddy")
print("file created")
file.close()
```

```python
#program on "r" mode
file=open("hvr.txt","r")
data=file.read()
print(data)
file.close()
```

```python
#progream on "a" mode
file=open("hvr.txt","a")
file.writelines(["\napple","\nname","\nend"])
```

```python
#program to read a single line
with open("hvr.txt","r") as file:
    line=file.readline()
    print(line)
```

```python
#program to read all lines
with open("hvr.txt","r") as file:
    lines=file.readlines()
    print(lines)
```

```python
with open("hvr.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(0)
  print(f.tell())
  print(f.read())
```

```python
#coping one file to another
with open("hvr.txt","r") as f1:
    data=f1.readlines()
with open("hvr1.txt","w") as f2:
    f2.writelines(data)
    print("data copied")
```

```python
#write a program to print each line of a file in revrse order
with open("hvr.txt","r") as f:
  data=f.readlines()
  for line in data:
    print(line[::-1])
```

```python
with open("hvr.txt","r") as f1:
    data=f1.read().split()
    data_lower=[i.lower() for i in data]
    sort_data=sorted(data_lower)
with open("hvr2.txt","w") as f2:
    for i in sort_data:
        f2.write(i+"\n")
print("done")

```

```python
#python program to compute the no of characters,words and lines in file
file=open("hvr.txt","r")
lines=file.readline()
char=0
words=0
line_count=len(lines)
for line in lines:
    wordslist=line.split()
    words=words+len(wordslist)
    char=char+len(line)
print("lines:",line_count,"words:",words,"characters:",char)
```

```python
  #list operations
l1=[1,5,10,'hello',20,'hii']
l2=[25,30,40]
#1.concatenation
print(l1+l2)
#2.repetition
print(l2*2)
#3.indexing
print(l1[1])
print(l1[-1])
#4.slicing
print(l1[:5])
print(l1[3:])
print(l1[::-1])
#5.membership operator
if 10 in l1:
  print("true")
else:
  print('false')
#6.iteration
for i in l1:
  print(i,end=" ")
```

```python
#built in functions
L1=[11,22,33,"harsha",44,55]
L2=[10,88,32,49]
#len()
print(len(L1))
#sum()
print(sum(L2))
#min()
print(min(L2))
#max()
print(max(L2))
#reversed()
print(list(reversed(L1)))
#sorted()
print(sorted(L2))
```

```python
#list methods
#1.append
l=[16,37,88,54,49]
l.append(60)
print(l)
#2.extend
l.extend(L2)
print(l)
#insert
l.insert(2,15)
print(l)
#pop
l.pop()
print(l)
#remove
l.remove(88)
print(l)
#count
print(l.count(49))
#sort
l.sort()
print(l)
#reverse
l.reverse()
print(l)
```

```python
#program to print sum of list elements with out using functions
l=[12,43,66,5,34,23,78,86]
sum=0
for i in l:
  sum+=i
print(sum)
```

```python
#print the list elements at odd index
l=[12,43,66,5,34,23,78,86]
for i in range(len(l)):
   if(i%2!=0):
     print(l[i],end=" ")
```

```python
#print the even numbers
l=[12,43,66,5,34,23,78,86]
for i in l:
  if(i%2==0):
    print(i,end=" ")
```

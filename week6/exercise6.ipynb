```python
#to print repeated elements
l = [int(i) for i in input().split()]
l2 = []
for i in l:
    if l.count(i) > 1:
        l2.append(i)
print(l2)
```

```python
#to print no repeated element
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)==1:
    l2.append(i)
print(l2)
```
